# Capa bronze: accidentes en carreteras

Carga el Excel de `data/raw` y lo guarda en Parquet en `data/bronze` sin limpiar ni transformar nada. Solo se agregan dos columnas de auditoría.

- **Entrada:** `data/raw/BD-Accidentes-Carreteras.xlsx`, hoja `registro_acidente`
- **Salida:** `data/bronze/accidentes.parquet`

La limpieza y el tipado se hacen en la capa silver.

## 1. Importar librerías y definir rutas

Las rutas son relativas a la carpeta `etl/`, donde vive este notebook.

```python
import pandas as pd
from pathlib import Path

origen = Path("../data/raw/BD-Accidentes-Carreteras.xlsx")
destino = Path("../data/bronze/accidentes.parquet")
```

## 2. Leer el Excel

Cada fila es un accidente. Se esperan 60.531 filas y 20 columnas.

```python
df = pd.read_excel(
    origen,
    sheet_name="registro_acidente"
)

df.shape
```

## 3. Agregar metadatos de carga

Indican cuándo y desde qué archivo se cargó el dato. El prefijo `_` las distingue de las columnas de la fuente.

```python
df["_fecha_carga"] = pd.Timestamp.now()
df["_archivo_origen"] = origen.name
```

## 4. Guardar en Parquet

Crea la carpeta `bronze` si no existe y sobrescribe el archivo en cada ejecución.

```python
destino.parent.mkdir(exist_ok=True)
df.to_parquet(destino, index=False)
```
