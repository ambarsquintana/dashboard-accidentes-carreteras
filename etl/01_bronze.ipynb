# Capa bronze: accidentes en carreteras

Ingesta del Excel fuente a Parquet, sin transformaciones.

- **Entrada:** `data/raw/BD-Accidentes-Carreteras.xlsx`, hoja `registro_acidente`
- **Salida:** `data/bronze/accidentes.parquet`
- **Grano:** una fila por accidente
- **Carga:** completa, sobrescribe el archivo en cada ejecución

## Configuración

```python
import pandas as pd
from pathlib import Path

origen  = Path("../data/raw/BD-Accidentes-Carreteras.xlsx")
destino = Path("../data/bronze/accidentes.parquet")
```

## Lectura

Esperado: 60.531 filas, 20 columnas.

```python
df = pd.read_excel(
    origen,
    sheet_name="registro_acidente"
)

df.shape
```

## Columnas de auditoría

```python
df["_fecha_carga"] = pd.Timestamp.now()
df["_archivo_origen"] = origen.name
```

## Validaciones

```python
assert len(df) > 0, "la fuente no tiene filas"
assert df.shape[1] == 22, "bronze no tiene las 20 columnas de la fuente más las 2 de auditoría"
assert df["id"].notna().all(), "hay filas sin id"
```

## Escritura

```python
destino.parent.mkdir(exist_ok=True)
df.to_parquet(destino, index=False)
```
